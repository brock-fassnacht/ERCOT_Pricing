```python
import os
import zipfile
import pandas as pd

# Directory containing the ZIP files
directory = 'ercot_dam_files'

# Initialize a list to store DataFrames
all_dataframes = []

# Loop through all files in the directory
for file_name in os.listdir(directory):
    # Construct the full file path
    file_path = os.path.join(directory, file_name)
    
    # Check if the file is a ZIP file
    if file_name.endswith('.zip'):
        with zipfile.ZipFile(file_path, 'r') as zip_ref:
            # List all files in the ZIP archive
            zip_file_list = zip_ref.namelist()
            
            # Process each Excel file in the ZIP archive
            for zip_file_name in zip_file_list:
                if zip_file_name.endswith(('.xls', '.xlsx')):  # Check for Excel files
                    with zip_ref.open(zip_file_name) as excel_file:
                        # Load the Excel file
                        excel_data = pd.ExcelFile(excel_file)
                        # Get sheet names
                        sheet_names = excel_data.sheet_names
                        print(f"File: {file_name}, Excel: {zip_file_name}, Sheets: {sheet_names}")
                        
                        # Read each sheet into a DataFrame
                        for sheet_name in sheet_names:
                            df = excel_data.parse(sheet_name)
                            # Add metadata columns for traceability
                            df['source_file'] = file_name  # ZIP file
                            df['excel_file'] = zip_file_name  # Excel file
                            df['sheet_name'] = sheet_name  # Sheet name
                            # Append to the list of DataFrames
                            all_dataframes.append(df)

# Combine all DataFrames into one if needed
final_dataframe = pd.concat(all_dataframes, ignore_index=True)

# Save or use final_dataframe as needed
print(final_dataframe.head())

```

```python
import requests
import yfinance as yf

# Define the API endpoint
url = " https://api.blockchain.info/charts/hash-rate?timespan=2years&rollingAverage=5days&format=json"

# Define your API key
api_key = "4f8d87fa-360b-4fd5-8335-ff5d62e08eb5"  # Replace with your actual API key

# Set up headers with the API key
headers = {
    "X-API-Token": api_key
}

try:
    # Send a GET request to the API
    response = requests.get(url, headers=headers)
    
    # Check if the request was successful
    if response.status_code == 200:
        # Parse the JSON data
        data = response.json()
        print("Data fetched successfully!")
        print(data)
    else:
        # Handle errors
        print(f"Failed to fetch data. Status code: {response.status_code}")
        print(f"Error message: {response.text}")
except Exception as e:
    print(f"An error occurred: {e}")


values = data['values']

# Convert to DataFrame
gnh = pd.DataFrame(values)

# Convert UNIX timestamp to datetime
gnh['x'] = pd.to_datetime(gnh['x'], unit='s')

# Rename columns for clarity
gnh.rename(columns={'x': 'Date', 'y': 'Hash Rate (TH/s)'}, inplace=True)

# Create ticker objects for Bitcoin and Ethereum
btc = yf.Ticker("BTC-USD")

# Fetch historical data
btc_history = btc.history(period="2y", interval="1h").reset_index()
```

```python
# Define the API endpoint
url = " https://api.blockchain.info/charts/transaction-fees?timespan=2years&rollingAverage=5days&format=json"

try:
    # Send a GET request to the API
    response = requests.get(url, headers=headers)
    
    # Check if the request was successful
    if response.status_code == 200:
        # Parse the JSON data
        data = response.json()
        print("Data fetched successfully!")
        print(data)
    else:
        # Handle errors
        print(f"Failed to fetch data. Status code: {response.status_code}")
        print(f"Error message: {response.text}")
except Exception as e:
    print(f"An error occurred: {e}")


# Extract values
values = data['values']

# Convert to DataFrame
tx = pd.DataFrame(values)

# Convert UNIX timestamp to datetime
tx['x'] = pd.to_datetime(tx['x'], unit='s')

# Rename columns for clarity
tx.rename(columns={'x': 'Date', 'y': 'Tx_Fees_BTC'}, inplace=True)
```

```python
btc_history["Date"] = pd.to_datetime(btc_history["Datetime"].dt.date)
btc_tbl = btc_history[["Datetime", "Date", "Close"]].merge(tx, on = "Date", how="left")
btc_tbl = btc_tbl.merge(gnh, on="Date", how='left')

btc_tbl = btc_tbl[~btc_tbl["Tx_Fees_BTC"].isnull()]

btc_tbl["EST_TX_Fees_Per_Block"] = btc_tbl["Tx_Fees_BTC"] / (24*6)

btc_tbl["Block_Reward"] = np.where(btc_tbl["Date"] < "04-20-2024", 6.25, 3.125)
btc_tbl["Total_Block_Reward"] = btc_tbl["Block_Reward"] + btc_tbl["EST_TX_Fees_Per_Block"] 

btc_tbl["Hash_Price_THs_day"] = 24*6*(btc_tbl["Total_Block_Reward"] * btc_tbl["Close"]) / btc_tbl["Hash Rate (TH/s)"]

btc_tbl.to_csv("BTC_Network_Metrics.csv")
```

```python
import os
import zipfile
import pandas as pd

# Directory containing the ZIP files
directory = "C:\\Users\\bfass\\OneDrive\\Desktop\\Fin tools\\ERCOT Energy Pricing\\RTM_Resource_Hubs\\downthemall"

# Initialize a list to store DataFrames
all_dataframeshub = []

# Loop through all files in the directory
for file_name in os.listdir(directory):
    # Construct the full file path
    file_path = os.path.join(directory, file_name)
    
    # Check if the file is a ZIP file
    if file_name.endswith('.zip') and '_xml' not in file_name:
        with zipfile.ZipFile(file_path, 'r') as zip_ref:
            # List all files in the ZIP archive
            zip_file_list = zip_ref.namelist()
            
            # Process each CSV file in the ZIP archive
            for zip_file_name in zip_file_list:
                if zip_file_name.endswith('.csv'):  # Check for CSV files
                    with zip_ref.open(zip_file_name) as csv_file:
                        # Load the CSV file into a DataFrame
                        df = pd.read_csv(csv_file)
                        # Add metadata columns for traceability
                        df['source_file'] = file_name  # ZIP file
                        df['csv_file'] = zip_file_name  # CSV file
                        # Append to the list of DataFrames
                        all_dataframeshub.append(df)

# Combine all DataFrames into one if needed
final_dataframe6 = pd.concat(all_dataframeshub, ignore_index=True)
```
